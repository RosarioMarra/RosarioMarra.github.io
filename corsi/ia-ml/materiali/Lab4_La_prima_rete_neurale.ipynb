## 1. Dati

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np, matplotlib.pyplot as plt

print("GPU:", tf.config.list_physical_devices("GPU"))
(x_tr, y_tr), (x_te, y_te) = keras.datasets.fashion_mnist.load_data()
x_tr, x_te = x_tr / 255.0, x_te / 255.0
classi = ["T-shirt", "Pantaloni", "Pullover", "Vestito", "Cappotto",
          "Sandalo", "Camicia", "Sneaker", "Borsa", "Stivaletto"]
plt.imshow(x_tr[0], cmap="gray"); plt.title(classi[y_tr[0]]); plt.show()
```

## 2. Rete densa

```python
model = keras.Sequential([
    keras.Input(shape=(28, 28)),
    layers.Flatten(),                         # 28×28 → 784 numeri
    layers.Dense(128, activation="relu"),     # strato nascosto
    layers.Dropout(0.2),                      # spegne il 20% dei neuroni a caso
    layers.Dense(10, activation="softmax"),   # 10 probabilità, una per classe
])
model.summary()
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
storia = model.fit(x_tr, y_tr, epochs=10, validation_split=0.1)

plt.plot(storia.history["loss"], label="training"); plt.plot(storia.history["val_loss"], label="validazione")
plt.xlabel("epoca"); plt.ylabel("loss"); plt.legend(); plt.show()
print("Accuratezza test rete densa:", model.evaluate(x_te, y_te, verbose=0)[1])
```

## 3. Rete convoluzionale

```python
cnn = keras.Sequential([
    keras.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation="relu"), layers.MaxPooling2D(),
    layers.Flatten(), layers.Dropout(0.3),
    layers.Dense(10, activation="softmax"),
])
cnn.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
cnn.fit(x_tr[..., None], y_tr, epochs=5, validation_split=0.1)
print("Accuratezza test CNN:", cnn.evaluate(x_te[..., None], y_te, verbose=0)[1])

pred = np.argmax(cnn.predict(x_te[:5, ..., None]), axis=1)
for p, r in zip(pred, y_te[:5]):
    print("previsto:", classi[p], "| reale:", classi[r])
```
